```python
trial_averaged_sub01_func1pt8mm = np.load("/media/ubuntu/sda/TrippleN/human_fMRI/trial_averaged_sub01_func1pt8mm.npz", allow_pickle=True)
newdatastim_73k = trial_averaged_sub01_func1pt8mm['newdatastim_73k']
response = trial_averaged_sub01_func1pt8mm['newdata']
response_shared1000 = response[:, :1000]
voxel_info = trial_averaged_sub01_func1pt8mm['voxel_info']
voxel_info = pd.DataFrame(voxel_info, columns=trial_averaged_sub01_func1pt8mm['roi_type_names'])
```

```python
out_dir = "/media/ubuntu/sda/TrippleN/human_fMRI"  # 你想保存的目录，按需要改
for k, arr in resp_by_region.items():
    name = regions[k]
    out_path = os.path.join(out_dir, f"resp_{name}_func1py8mm.npy")
    np.save(out_path, arr)
    print("saved:", out_path)
```
