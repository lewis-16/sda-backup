```python
from mne.io import concatenate_raws, read_raw_edf
import matplotlib.pyplot as plt
import mne
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
raw.load_data()
raw.filter(l_freq=0.5, h_freq=600)
```

```python
events_from_annot, event_dict = mne.events_from_annotations(sEEG_raw)
```

```python
# 删除 event_dict 中 key 包含 '000000' 的项
event_dict_filtered = {k: v for k, v in event_dict.items() if '000000' not in str(k)}
print(f"原始 event_dict 项数: {len(event_dict)}")
print(f"过滤后 event_dict 项数: {len(event_dict_filtered)}")
print(f"删除的项: {[k for k in event_dict.keys() if '000000' in str(k)]}")

# 获取过滤后的 event_dict 中的所有 event ID
valid_event_ids = set(event_dict_filtered.values())
print(f"\n有效的 event ID: {valid_event_ids}")

# 过滤 events_from_annot，只保留 event_id 在过滤后的 event_dict 中的事件
events_from_annot_filtered = events_from_annot[np.isin(events_from_annot[:, 2], list(valid_event_ids))]
print(f"\n原始 events_from_annot 数量: {len(events_from_annot)}")
print(f"过滤后 events_from_annot 数量: {len(events_from_annot_filtered)}")

# 更新 event_dict 和 events_from_annot
event_dict = event_dict_filtered
events_from_annot = events_from_annot_filtered
```

```python
event_dict
```

```python
from matplotlib.backends.backend_pdf import PdfPages
from matplotlib import gridspec

# 获取采样率和数据
sfreq = sEEG_raw.info['sfreq']
n_channels = len(sEEG_raw.ch_names)
print(f"采样率: {sfreq} Hz")
print(f"通道数: {n_channels}")
print(f"事件数: {len(events_from_annot)}")

# 创建event_id到描述的映射（反转event_dict）
id_to_description = {v: k for k, v in event_dict.items()}

# 时间窗口设置（样本点）
t_before = 5000  # event前5000个样本点
t_after = 50000  # event后50000个样本点

# 创建PDF文件
pdf_filename = 'sEEG_events_traces.pdf'
with PdfPages(pdf_filename) as pdf:
    # 遍历每个事件
    for event_idx, event in enumerate(events_from_annot):
        event_sample = event[0]  # 事件在原始数据中的样本点位置
        event_id = event[2]  # 事件ID
        
        # 获取事件描述
        event_description = id_to_description.get(event_id, f'Unknown (ID: {event_id})')
        # 如果是numpy字符串类型，转换为普通字符串
        if isinstance(event_description, np.str_):
            event_description = str(event_description)
        
        # 计算时间窗口的起始和结束样本点
        start_sample = max(0, event_sample - t_before)
        end_sample = min(sEEG_raw.n_times, event_sample + t_after)
        
        # 提取数据
        data, times = sEEG_raw[:, int(start_sample):int(end_sample)]
        
        # 计算所有通道的全局最小值和最大值（用于统一scale）
        data_min = np.min(data)
        data_max = np.max(data)
        # 添加一些边距
        y_margin = (data_max - data_min) * 0.05
        y_min = data_min - y_margin
        y_max = data_max + y_margin
        
        # 调整event在提取数据中的位置（样本点索引）
        actual_event_idx = event_sample - start_sample
        
        # 创建图形（每页一个事件）
        fig = plt.figure(figsize=(20, 30))
        gs = gridspec.GridSpec(n_channels, 1, hspace=0.3)
        
        # 为每个通道绘制子图
        for ch_idx, ch_name in enumerate(sEEG_raw.ch_names):
            ax = fig.add_subplot(gs[ch_idx, 0])
            
            # 绘制trace
            ax.plot(times, data[ch_idx, :], linewidth=0.5, color='black')
            
            # 在event时间点绘制竖线
            event_time = times[actual_event_idx] if actual_event_idx < len(times) else times[-1]
            ax.axvline(x=event_time, color='red', linestyle='--', linewidth=1, alpha=0.7)
            
            # 设置标签和格式
            ax.set_ylabel(ch_name, rotation=0, ha='right', va='center', fontsize=6)
            ax.set_xlim(times[0], times[-1])
            # 统一所有子图的y轴范围
            ax.set_ylim(y_min, y_max)
            ax.tick_params(labelsize=5)
            ax.grid(True, alpha=0)
            ax.set_yticks([])
            
            # 只在最后一个子图显示x轴标签
            if ch_idx < n_channels - 1:
                ax.set_xticklabels([])
            else:
                ax.set_xlabel('Time (s)', fontsize=8)
        
        # 添加标题（显示事件描述而不是ID）
        fig.suptitle(f'Event {event_idx + 1}/{len(events_from_annot)} - {event_description} - Sample: {event_sample}', 
                     fontsize=12, y=0.995)
        
        # 保存到PDF
        pdf.savefig(fig, bbox_inches='tight', dpi=150)
        plt.close(fig)
        
        if (event_idx + 1) % 10 == 0:
            print(f"已处理 {event_idx + 1}/{len(events_from_annot)} 个事件")

print(f"\nPDF文件已保存: {pdf_filename}")
```

```python
# Plot PDF for each seizure from start to end
# Find all event_id=20 (seizure start) and event_id=19 (end) event pairs

# Get event_id to description mapping
id_to_description = {v: k for k, v in event_dict.items()}
seizure_event_id = 20  # Seizure start
end_event_id = 19      # End

# Find all seizure start and end events
seizure_starts = events_from_annot[events_from_annot[:, 2] == seizure_event_id]
end_events = events_from_annot[events_from_annot[:, 2] == end_event_id]

print(f"Found {len(seizure_starts)} seizure start events")
print(f"Found {len(end_events)} end events")

# Find corresponding end event for each seizure start
seizure_pairs = []
for seizure_start in seizure_starts:
    start_sample = seizure_start[0]
    # Find the nearest end event after seizure start
    end_candidates = end_events[end_events[:, 0] > start_sample]
    if len(end_candidates) > 0:
        end_sample = end_candidates[0, 0]  # Take the first end event
        seizure_pairs.append((start_sample, end_sample))
        print(f"Seizure starts at sample {start_sample}, ends at sample {end_sample}, duration: {(end_sample - start_sample) / sfreq:.2f} s")
    else:
        print(f"Warning: Seizure at sample {start_sample} has no corresponding end event")

print(f"\nFound {len(seizure_pairs)} complete seizure-end pairs")

# Create PDF file
pdf_filename = 'sEEG_seizure_to_end_traces.pdf'
n_channels = len(sEEG_raw.ch_names)

with PdfPages(pdf_filename) as pdf:
    # Iterate through each seizure-end pair
    for pair_idx, (start_sample, end_sample) in enumerate(seizure_pairs):
        # Extract data (from seizure start to end, extended by 1000 samples on each side)
        extend_samples = 1000
        extract_start = max(0, start_sample - extend_samples)
        extract_end = min(sEEG_raw.n_times, end_sample + extend_samples)
        data, times = sEEG_raw[:, int(extract_start):int(extract_end)]
        
        # Calculate sample indices for seizure start and end in extracted data
        start_idx_in_data = start_sample - extract_start
        end_idx_in_data = end_sample - extract_start
        
        # Get corresponding actual time values
        start_time = times[start_idx_in_data] if start_idx_in_data < len(times) else times[0]
        end_time = times[end_idx_in_data] if end_idx_in_data < len(times) else times[-1]
        
        # Calculate global min and max across all channels (for unified scale)
        data_min = np.min(data)
        data_max = np.max(data)
        # Add some margin
        y_margin = (data_max - data_min) * 0.05
        y_min = data_min - y_margin
        y_max = data_max + y_margin
        
        # Create figure (one page per seizure)
        fig = plt.figure(figsize=(20, 30))
        gs = gridspec.GridSpec(n_channels, 1, hspace=0.3)
        
        # Plot subplot for each channel
        for ch_idx, ch_name in enumerate(sEEG_raw.ch_names):
            ax = fig.add_subplot(gs[ch_idx, 0])
            
            # Plot trace
            ax.plot(times, data[ch_idx, :], linewidth=0.5, color='black')
            
            # Draw vertical lines at seizure start and end time points
            ax.axvline(x=start_time, color='green', linestyle='--', linewidth=1.5, alpha=0.8, label='Seizure Start')
            ax.axvline(x=end_time, color='red', linestyle='--', linewidth=1.5, alpha=0.8, label='Seizure End')
            
            # Set labels and formatting
            ax.set_ylabel(ch_name, rotation=0, ha='right', va='center', fontsize=6)
            ax.set_xlim(times[0], times[-1])
            # Unified y-axis range for all subplots
            ax.set_ylim(y_min, y_max)
            ax.grid(False)
            ax.set_xticks([])
            ax.set_yticks([])
            
            # Only show x-axis label and legend in the last subplot
            if ch_idx < n_channels - 1:
                ax.set_xticklabels([])
            else:
                ax.set_xlabel('Time (s)', fontsize=8)
                ax.legend(loc='upper right', fontsize=6)
        
        # Add title (English)
        duration = (end_sample - start_sample) / sfreq
        fig.suptitle(f'Seizure {pair_idx + 1}/{len(seizure_pairs)} - Start Sample: {start_sample} - End Sample: {end_sample} - Duration: {duration:.2f} s', 
                     fontsize=12, y=0.995)
        
        # Save to PDF
        pdf.savefig(fig, bbox_inches='tight', dpi=150)
        plt.close(fig)
        
        print(f"Processed seizure {pair_idx + 1}/{len(seizure_pairs)}")

print(f"\nPDF file saved: {pdf_filename}")
```

```python
# Calculate correlation matrices for each seizure
# Compute channel-to-channel correlation for each seizure period

# Store correlation matrices for each seizure
correlation_matrices = []

n_channels = len(sEEG_raw.ch_names)
print(f"Number of channels: {n_channels}")
print(f"Number of seizures: {len(seizure_pairs)}\n")

# Calculate correlation matrix for each seizure
for seizure_idx, (start_sample, end_sample) in enumerate(seizure_pairs):
    # Extract data only during the seizure period (without extension)
    seizure_data, _ = sEEG_raw[:, int(start_sample):int(end_sample)]
    
    # Calculate correlation matrix using numpy (much faster)
    # Each row is a channel, each column is a time point
    # np.corrcoef computes correlation between rows (channels)
    corr_matrix = np.corrcoef(seizure_data)
    
    correlation_matrices.append(corr_matrix)
    
    # Print statistics (using upper triangle to avoid counting diagonal and duplicates)
    triu_indices = np.triu_indices(n_channels, k=1)
    corr_values = corr_matrix[triu_indices]
    
    print(f"Seizure {seizure_idx + 1}:")
    print(f"  Duration: {(end_sample - start_sample) / sfreq:.2f} s")
    print(f"  Correlation matrix shape: {corr_matrix.shape}")
    print(f"  Mean correlation: {np.mean(corr_values):.4f}")
    print(f"  Std correlation: {np.std(corr_values):.4f}")
    print(f"  Min correlation: {np.min(corr_values):.4f}")
    print(f"  Max correlation: {np.max(corr_values):.4f}\n")

print(f"Successfully computed {len(correlation_matrices)} correlation matrices")
print(f"\nCorrelation matrices are stored in 'correlation_matrices' list")
print(f"Access individual matrices: correlation_matrices[0], correlation_matrices[1], etc.")
print(f"\nEach matrix is {n_channels}x{n_channels} and contains Pearson correlation coefficients")
```

```python
# Plot heatmaps for each correlation matrix as PDF
# Create separate heatmaps for each seizure with channel names, one page per seizure

from matplotlib.backends.backend_pdf import PdfPages

# Get channel names
channel_names = sEEG_raw.ch_names
n_channels = len(channel_names)

# Create PDF file
pdf_filename = 'correlation_matrices_heatmaps.pdf'

with PdfPages(pdf_filename) as pdf:
    # Plot heatmap for each seizure (one page per seizure)
    for seizure_idx in range(len(correlation_matrices)):
        # Create figure for this seizure
        fig, ax = plt.subplots(figsize=(30, 30))
        
        # Get correlation matrix
        corr_matrix = correlation_matrices[seizure_idx]
        
        # Create heatmap
        im = ax.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1, aspect='auto')
        
        # Set ticks and labels
        # Since there are 134 channels, we'll show all channel names but may need to rotate them
        ax.set_xticks(np.arange(n_channels))
        ax.set_yticks(np.arange(n_channels))
        ax.set_xticklabels(channel_names, rotation=90, fontsize=4, ha='right')
        ax.set_yticklabels(channel_names, fontsize=4)
        
        # Add colorbar
        cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
        cbar.set_label('Correlation Coefficient', fontsize=12)
        
        # Add title
        duration = (seizure_pairs[seizure_idx][1] - seizure_pairs[seizure_idx][0]) / sfreq
        start_sample = seizure_pairs[seizure_idx][0]
        end_sample = seizure_pairs[seizure_idx][1]
        ax.set_title(f'Seizure {seizure_idx + 1}/{len(correlation_matrices)} - Duration: {duration:.2f} s\nStart Sample: {start_sample} - End Sample: {end_sample}', 
                     fontsize=16, pad=20)
        
        # Add grid for better readability (optional, can be removed if too cluttered)
        ax.set_xticks(np.arange(n_channels) - 0.5, minor=True)
        ax.set_yticks(np.arange(n_channels) - 0.5, minor=True)
        ax.grid(which='minor', color='gray', linestyle='-', linewidth=0.1)
        
        # Add labels
        ax.set_xlabel('Channels', fontsize=12)
        ax.set_ylabel('Channels', fontsize=12)
        
        plt.tight_layout()
        
        # Save to PDF
        pdf.savefig(fig, bbox_inches='tight', dpi=150)
        plt.close(fig)
        
        print(f"Processed seizure {seizure_idx + 1}/{len(correlation_matrices)}")

print(f"\nPDF file saved: {pdf_filename}")
```

```python
# Plot time-frequency representations for each seizure
# Frequency range: 0.5-500 Hz

from mne.time_frequency import tfr_morlet

# Frequency parameters
freqs = np.logspace(np.log10(0.5), np.log10(500), 50)  # Logarithmic spacing from 0.5 to 500 Hz
n_cycles = freqs / 2.0  # Number of cycles (adjust based on frequency)

# Create PDF file for time-frequency plots
pdf_filename = 'seizure_time_frequency.pdf'

with PdfPages(pdf_filename) as pdf:
    for seizure_idx, (start_sample, end_sample) in enumerate(seizure_pairs):
        # Extract data during seizure period (with extension for context)
        extend_samples = 1000
        extract_start = max(0, start_sample - extend_samples)
        extract_end = min(sEEG_raw.n_times, end_sample + extend_samples)
        
        # Create a temporary Raw object for this seizure period
        seizure_raw = sEEG_raw.copy().crop(tmin=extract_start/sfreq, tmax=extract_end/sfreq)
        
        # Calculate time-frequency representation using Morlet wavelets
        # Average across all channels for overall spectral activity
        tfr = tfr_morlet(seizure_raw, freqs=freqs, n_cycles=n_cycles, 
                         use_fft=True, return_itc=False, decim=1, n_jobs=1)
        
        # Average across channels
        tfr_avg = tfr.average()
        
        # Create figure
        fig, ax = plt.subplots(figsize=(20, 10))
        
        # Plot time-frequency representation
        tfr_avg.plot(axes=ax, colorbar=True, show=False, 
                     vmin=None, vmax=None, cmap='viridis')
        
        # Add vertical lines for seizure start and end
        seizure_start_time = (start_sample - extract_start) / sfreq
        seizure_end_time = (end_sample - extract_start) / sfreq
        ax.axvline(x=seizure_start_time, color='green', linestyle='--', 
                   linewidth=2, alpha=0.8, label='Seizure Start')
        ax.axvline(x=seizure_end_time, color='red', linestyle='--', 
                   linewidth=2, alpha=0.8, label='Seizure End')
        
        # Set title
        duration = (end_sample - start_sample) / sfreq
        ax.set_title(f'Time-Frequency: Seizure {seizure_idx + 1}/{len(seizure_pairs)} - Duration: {duration:.2f} s\n'
                     f'Start Sample: {start_sample} - End Sample: {end_sample} - Frequency: 0.5-500 Hz',
                     fontsize=14, pad=20)
        
        # Set frequency axis to log scale for better visualization
        ax.set_yscale('log')
        ax.set_ylabel('Frequency (Hz)', fontsize=12)
        ax.set_xlabel('Time (s)', fontsize=12)
        ax.legend(loc='upper right', fontsize=10)
        
        plt.tight_layout()
        
        # Save to PDF
        pdf.savefig(fig, bbox_inches='tight', dpi=150)
        plt.close(fig)
        
        print(f"Processed seizure {seizure_idx + 1}/{len(seizure_pairs)}")

print(f"\nPDF file saved: {pdf_filename}")
```
