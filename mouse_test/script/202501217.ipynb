```python
# 读取 P_ASD 目录下的所有日期目录，按日期顺序处理
base_dir = Path("/media/ubuntu/sda/mouse_test/raw_data/P_ASD")

# 获取所有日期目录
date_dirs = [d for d in base_dir.iterdir() if d.is_dir()]
# 按目录名排序（目录名中包含日期，如 251202, 251203 等）
date_dirs = sorted(date_dirs, key=lambda x: x.name)

print(f"找到 {len(date_dirs)} 个日期目录:")
for dd in date_dirs:
    print(f"  - {dd.name}")

# 收集所有需要读取的文件路径
all_rhd_files = []

for date_dir in date_dirs:
    # 获取该日期目录下的所有 .rhd 文件
    rhd_files = list(date_dir.glob("*.rhd"))
    
    # 排除文件名中包含 "6" 的文件（根据需求调整）
    # 如果需要排除其他模式，可以修改这里的条件
    rhd_files = [f for f in rhd_files if "6" not in f.name]
    
    # 按文件名排序（确保同一日期下的文件按顺序读取）
    rhd_files = sorted(rhd_files)
    
    print(f"\n日期目录 {date_dir.name}:")
    print(f"  找到 {len(rhd_files)} 个 .rhd 文件（已排除包含'6'的文件）")
    
    all_rhd_files.extend(rhd_files)

print(f"\n总共需要读取 {len(all_rhd_files)} 个文件")

# 读取所有文件
recording_raw_list = []
for i, rhd_file in enumerate(all_rhd_files):
    print(f"读取文件 {i+1}/{len(all_rhd_files)}: {rhd_file.name}")
    recording_raw_list.append(se.read_intan(str(rhd_file), stream_id='0'))

# 合并所有 recording
print("\n合并所有 recording...")
recording_raw = concatenate_recordings(recording_list=recording_raw_list)
recording_raw = recording_raw.select_channels(channel_list)

# 预处理
recording_raw = spre.unsigned_to_signed(recording_raw)
recording_recorded = spre.bandpass_filter(recording_raw, freq_min=300, freq_max=3000)
recording_recorded = spre.notch_filter(recording_recorded, freq=50)
recording_f = spre.common_reference(recording_recorded, reference="global", operator="median")

probe = read_probeinterface('/media/ubuntu/sda/mouse_test/probe/tip_probe_128_1.json')
recording_f = recording_f.set_probegroup(probe)

output_folder = '/media/ubuntu/sda/mouse_test/sorted/P_ASD_combined'
os.makedirs(output_folder, exist_ok=True)

# recording_preprocessed = recording_f.save(format="binary", n_jobs=5)

# sorting_kilosort4 = ss.run_sorter(
#     sorter_name="kilosort4", 
#     recording=recording_preprocessed, 
#     folder=output_folder + "/kilosort4"
# )

# analyzer_kilosort4 = si.create_sorting_analyzer(
#     sorting=sorting_kilosort4, 
#     recording=recording_preprocessed, 
#     format='binary_folder', 
#     folder=output_folder + '/analyzer_kilosort4_binary'
# )

# extensions_to_compute = [
#     "random_spikes",
#     "waveforms",
#     "noise_levels",
#     "templates",
#     "unit_locations",
#     "spike_locations",
#     "correlograms",
#     "template_similarity"
# ]

# extension_params = {
#     "unit_locations": {"method": "center_of_mass"},
#     "spike_locations": {"ms_before": 0.1},
#     "correlograms": {"bin_ms": 0.1},
#     "template_similarity": {"method": "cosine_similarity"}
# }

# analyzer_kilosort4.compute(extensions_to_compute, extension_params=extension_params, n_jobs=10)

# qm_params = sqm.get_default_qm_params()
# analyzer_kilosort4.compute("quality_metrics", qm_params, n_jobs=10)

# sexp.export_to_phy(analyzer_kilosort4, output_folder + "/phy_folder_for_kilosort", verbose=True)
```

```python
recording_preprocessed = recording_f.save(format="binary", n_jobs=30)

sorting_kilosort4 = ss.run_sorter(
    sorter_name="kilosort4", 
    recording=recording_preprocessed, 
    folder=output_folder + "/kilosort4"
)

analyzer_kilosort4 = si.create_sorting_analyzer(
    sorting=sorting_kilosort4, 
    recording=recording_preprocessed, 
    format='binary_folder', 
    folder=output_folder + '/analyzer_kilosort4_binary'
)

extensions_to_compute = [
    "random_spikes",
    "waveforms",
    "noise_levels",
    "templates",
    "unit_locations",
    "spike_locations",
    "correlograms",
    "template_similarity"
]+

extension_params = {
    "unit_locations": {"method": "center_of_mass"},
    "spike_locations": {"ms_before": 0.1},
    "correlograms": {"bin_ms": 0.1},
    "template_similarity": {"method": "cosine_similarity"}
}

analyzer_kilosort4.compute(extensions_to_compute, extension_params=extension_params, n_jobs=30)

qm_params = sqm.get_default_qm_params()
analyzer_kilosort4.compute("quality_metrics", qm_params, n_jobs=30)

sexp.export_to_phy(analyzer_kilosort4, output_folder + "/phy_folder_for_kilosort", verbose=True, n_jobs=30)
```
