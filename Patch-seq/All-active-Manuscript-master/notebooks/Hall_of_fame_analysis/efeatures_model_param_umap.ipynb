# Ephys, Parameter embedding - Experiment and Model

```python
import pandas as pd
from ateamopt.utils import utility
import matplotlib.colors as mpl_colors
import numpy as np
import man_opt.utils as man_utils
import os
from bokeh.layouts import gridplot
from bokeh.models import CDSView, ColumnDataSource, Legend, BooleanFilter
from bokeh.plotting import figure, show
from bokeh.transform import factor_cmap, factor_mark
from bokeh.resources import INLINE
import bokeh.io
from bokeh.io import export_png
import pyarrow.feather as feather

bokeh.io.output_notebook(INLINE)
```

## Utility Functions

```python
def intLinspace(elements, numElems):
    idx = np.round(np.linspace(0, len(elements) - 1, numElems)).astype(int)
    return [elements[id_] for id_ in idx]
```

## Get the data

```python
data_path = os.path.join(os.getcwd(), os.pardir, os.pardir, 'assets', 'aggregated_data')
mouse_data_filename = os.path.join(data_path, 'Mouse_class_data.csv')
mouse_datatype_filename = os.path.join(data_path, 'Mouse_class_datatype.csv')
train_ephys_max_amp_filename = os.path.join(data_path, 'train_ephys_max_amp.csv')
train_ephys_max_amp_dtype_filename = os.path.join(data_path, 'train_ephys_max_amp_dtype.csv')
train_ephys_max_amp_fields_filename = os.path.join(data_path, 'train_ephys_max_amp_fields.json')
hof_model_ephys_max_amp_filename = os.path.join(data_path, 'hof_model_ephys_max_amp.csv')
hof_model_ephys_max_amp_dtype_filename = os.path.join(data_path, 'hof_model_ephys_max_amp_datatype.csv')
annotation_datapath = os.path.join(data_path, 'anno.feather')
me_ttype_map_path = os.path.join(data_path, 'me_ttype.pkl')
exp_feature_embedding_path = os.path.join(data_path, 'efeat_exp_umap_embedding.csv')
model_feature_embedding_path = os.path.join(data_path, 'efeat_model_umap_embedding.csv')
param_embedding_path = os.path.join(data_path, 'param_umap_embedding.csv')


mouse_data_df = man_utils.read_csv_with_dtype(mouse_data_filename, mouse_datatype_filename)
ephys_data = man_utils.read_csv_with_dtype(train_ephys_max_amp_filename, train_ephys_max_amp_dtype_filename)
model_ephys_data = man_utils.read_csv_with_dtype(hof_model_ephys_max_amp_filename,
                                                 hof_model_ephys_max_amp_dtype_filename)
exp_feat_embedding = pd.read_csv(exp_feature_embedding_path, index_col=None)
hof_efeat_embedding = pd.read_csv(model_feature_embedding_path, index_col=None)
hof_param_embedding = pd.read_csv(param_embedding_path, index_col=None)

anno_data = feather.read_feather(annotation_datapath)
me_cluster = mouse_data_df.loc[mouse_data_df.hof_index == 0, ['Cell_id', 'me_type']]
me_ttype_map = utility.load_pickle(me_ttype_map_path)
me_ttype_map = {**me_ttype_map, 'Other': 'Other'}
```

## Superficial vs Deep PCs

```python
def create_layer_cty_for_PCs(df, target_field):
    cty_cluster = mouse_data_df.loc[mouse_data_df.hof_index == 0, ['Cell_id', 'Layer', 'Broad_Cre_line']]
    cty_cluster[target_field] = cty_cluster.apply(
    lambda x: 'L%s PC'%x['Layer'] if x['Broad_Cre_line'] == 'Pyr' else 'Other' , axis=1)
    cty_cluster[target_field] = cty_cluster[target_field].apply(lambda x: x if x in ['L2/3 PC', 'L5 PC'] else 'Other')
    cty_cluster = cty_cluster.drop(columns=['Layer', 'Broad_Cre_line'])
    df['Cell_id'] = df['Cell_id'].astype(str)
    return pd.merge(df, cty_cluster, how='left', on='Cell_id')
```

```python
least_pop_index = 6
hof_num = 40
exp_feat_layer = create_layer_cty_for_PCs(exp_feat_embedding, 'cty_layer')
hof_efeat_layer = create_layer_cty_for_PCs(hof_efeat_embedding, 'cty_layer')
hof_param_layer = create_layer_cty_for_PCs(hof_param_embedding, 'cty_layer')

exp_feat_layer['data_type'] = 'exp features'
hof_efeat_layer['data_type'] = 'model features'
hof_param_layer['data_type'] = 'model parameters'
col_list = ['Cell_id', 'x-dim', 'y-dim', 'data_type', 'cty_layer']
df_list = [exp_feat_layer, hof_efeat_layer, hof_param_layer]

concat_df = []
for df_ in df_list:
    concat_df.append(df_)
concat_df = pd.concat(concat_df, axis=0)
```

## Visualize the transform

```python
source_data = ColumnDataSource(data=concat_df)

exp_feat_filter = [True if data_type == 'exp features' else False for data_type in source_data.data['data_type']]
exp_feat_view = CDSView(source=source_data, filters=[BooleanFilter(exp_feat_filter)])
model_feat_filter = [True if data_type == 'model features' else False for data_type in source_data.data['data_type']]
model_feat_view = CDSView(source=source_data, filters=[BooleanFilter(model_feat_filter)])
model_param_filter = [True if data_type == 'model parameters' else False for data_type in source_data.data['data_type']]
model_param_view = CDSView(source=source_data, filters=[BooleanFilter(model_param_filter)])

highlight_cty = ['L2/3 PC', 'L5 PC']
cty_order = exp_feat_layer.cty_layer.unique().tolist()
cty_order.sort()
cty_colors = {'L2/3 PC': '#90d743', 'L5 PC': '#443983', 'Other': '#D3D3D30D'}
filter_colors = [cty_colors[cty_] if cty_ in highlight_cty else mpl_colors.to_hex('lightgrey') for cty_ in cty_order]
markers = ['circle' if cty_ in highlight_cty else 'dot' for cty_ in cty_order]
```

```python
x_var = "x-umap"
y_var = "y-umap"

plot_size_and_tools = {'plot_height': 375, 'plot_width': 300,
                        'tools':['box_select', 'reset', 'save', 'help']}
p_exp = figure(title="Experiment Features", plot_height=375, plot_width=400)
p_exp.add_layout(Legend(), 'left')
p_exp.scatter(x_var, y_var, source=source_data, view=exp_feat_view, legend_field="cty_layer", fill_alpha=1, size=8,
          marker=factor_mark('cty_layer', markers, cty_order),
          color=factor_cmap('cty_layer', palette=filter_colors, factors=cty_order))

p_best = figure(title="Model Features", **plot_size_and_tools)
p_best.scatter(x_var, y_var, source=source_data, view=model_feat_view, legend_field="cty_layer", fill_alpha=1, size=4,
           marker=factor_mark('cty_layer', markers, cty_order),
          color=factor_cmap('cty_layer', palette=filter_colors, factors=cty_order))

p_hof = figure(title="Model Parameters", **plot_size_and_tools)

p_hof.scatter(x_var, y_var, source=source_data, view=model_param_view, legend_field="cty_layer", fill_alpha=1, size=4,
          marker=factor_mark('cty_layer', markers, cty_order),
          color=factor_cmap('cty_layer', palette=filter_colors, factors=cty_order))

p_grid = [p_exp, p_best, p_hof]
for i, p_ in enumerate(p_grid):
    if i != 0:
        p_.legend.visible = False
    if i == 0:
        p_.legend.label_text_font_size = "12pt"
        p_.yaxis.axis_label = y_var
    p_.xaxis.axis_label = x_var
    p_.xgrid.visible = False
    p_.ygrid.visible = False
    p_.axis.visible = False
    p_.title.text_font_size = '12pt'


p = gridplot([p_grid])
show(p)

utility.create_dirpath('./figures/')
export_png(p, filename='figures/umap_cty_layerwise.png')
```

```python
def filter_for_ttype(df, select_ttypes):
    me_cluster = mouse_data_df.loc[mouse_data_df.hof_index == 0, [
    'Cell_id', 'Broad_Cre_line', 'me_type']]
    me_cluster['me_type'] = me_cluster['me_type'].fillna('Other')
    me_cluster['ttype'] = me_cluster['me_type'].apply(lambda x: me_ttype_map[x])
    me_cluster['ttype'] = me_cluster['ttype'].apply(lambda x: x if x in select_ttypes else 'Other')
    df = df.drop(columns='ttype')
    df['Cell_id'] = df['Cell_id'].astype(str)
    return pd.merge(df, me_cluster, on='Cell_id', how='left')

select_ttypes = ['L5 IT', 'L5 PT']
exp_feat_ttype = filter_for_ttype(exp_feat_embedding, select_ttypes)
hof_efeat_ttype = filter_for_ttype(hof_efeat_embedding, select_ttypes)
hof_param_ttype = filter_for_ttype(hof_param_embedding, select_ttypes)

highlight_cty = select_ttypes

exp_feat_ttype['data_type'] = 'exp features'
hof_efeat_ttype['data_type'] = 'model features'
hof_param_ttype['data_type'] = 'model parameters'
col_list = ['Cell_id', 'x-dim', 'y-dim', 'data_type', 'cty_layer']
df_list = [exp_feat_ttype, hof_efeat_ttype, hof_param_ttype]

concat_df = []
for df_ in df_list:
    concat_df.append(df_)
concat_df = pd.concat(concat_df, axis=0)

ttypes = sorted(exp_feat_ttype.ttype.unique().tolist())
tasic_colors = {'L5 PT':'#008A61', 'L5 IT':'#0D5B78', 'Other':'#D3D3D3'}
filter_colors = [tasic_colors[cty_] for cty_ in ttypes]
markers = ['circle' if cty_ in highlight_cty else 'dot' for cty_ in ttypes]
```

```python
source_data = ColumnDataSource(data=concat_df)
exp_feat_filter = [True if data_type == 'exp features' else False for data_type in source_data.data['data_type']]
exp_feat_view = CDSView(source=source_data, filters=[BooleanFilter(exp_feat_filter)])
model_feat_filter = [True if data_type == 'model features' else False for data_type in source_data.data['data_type']]
model_feat_view = CDSView(source=source_data, filters=[BooleanFilter(model_feat_filter)])
model_param_filter = [True if data_type == 'model parameters' else False for data_type in source_data.data['data_type']]
model_param_view = CDSView(source=source_data, filters=[BooleanFilter(model_param_filter)])

p_exp = figure(title="Experiment Features", plot_height=375, plot_width=400, tools=plot_size_and_tools['tools'])
p_exp.add_layout(Legend(), 'left')
p_exp.scatter(x_var, y_var, source=source_data, view=exp_feat_view, legend_field="ttype", fill_alpha=.8, size=8,
          marker=factor_mark('ttype', markers, ttypes),
          color=factor_cmap('ttype', palette=filter_colors, factors=ttypes))

p_best = figure(title="Model Features", **plot_size_and_tools)
p_best.scatter(x_var, y_var, source=source_data, view=model_feat_view, legend_field="ttype", fill_alpha=.8, size=4,
              marker=factor_mark('ttype', markers, ttypes),
               color=factor_cmap('ttype', palette=filter_colors, factors=ttypes))

p_hof = figure(title="Model Parameters", **plot_size_and_tools)


p_hof.scatter(x_var, y_var, source=source_data, view=model_param_view, legend_field="ttype", fill_alpha=.8, size=4,
            marker=factor_mark('ttype', markers, ttypes),
          color=factor_cmap('ttype', palette=filter_colors, factors=ttypes))

p_grid = [p_exp, p_best, p_hof]
for i, p_ in enumerate(p_grid):
    if i != 0:
        p_.legend.visible = False
    if i == 0:
        p_.legend.label_text_font_size = "12pt"
        p_.yaxis.axis_label = y_var
    p_.xaxis.axis_label = x_var
    p_.xgrid.visible = False
    p_.ygrid.visible = False
    p_.axis.visible = False
    p_.title.text_font_size = '12pt'
    
p = gridplot([p_grid])

show(p)
export_png(p, filename='figures/umap_ttype.png')
```
